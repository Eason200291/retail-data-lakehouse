```python
clean_df = spark.table("clean_sales")
```

```python
from pyspark.sql.functions import sum, round

top_products_revenue = (
    clean_df
    .groupBy("Description")
    .agg(
        round(sum("Revenue"), 2).alias("TotalRevenue")
    )
    .orderBy("TotalRevenue", ascending=False)
    .limit(10)
)

display(top_products_revenue)
```

```python
from pyspark.sql.functions import sum

top_products_quantity = (
    clean_df
    .groupBy("Description")
    .agg(
        sum("Quantity").alias("TotalQuantity")
    )
    .orderBy("TotalQuantity", ascending=False)
    .limit(10)
)

display(top_products_quantity)
```

```python
from pyspark.sql.functions import sum, round

top_customers_revenue = (
    clean_df
    .groupBy("CustomerID")
    .agg(
        round(sum("Revenue"), 2).alias("TotalRevenue")
    )
    .orderBy("TotalRevenue", ascending=False)
    .limit(10)
)

display(top_customers_revenue)
```

```python
from pyspark.sql.functions import sum, round

country_revenue = (
    clean_df
    .groupBy("Country")
    .agg(
        round(sum("Revenue"), 2).alias("TotalRevenue")
    )
    .orderBy("TotalRevenue", ascending=False)
)

display(country_revenue)
```

```python
from pyspark.sql.functions import date_format, sum, round

monthly_df = (
    clean_df
    .withColumn(
        "Month",
        date_format("InvoiceDate", "yyyy-MM")
    )
)

monthly_revenue = (
    monthly_df
    .groupBy("Month")
    .agg(
        sum("Revenue").alias("TotalRevenue")
    )
    .orderBy("Month")
)

display(monthly_revenue)
```
