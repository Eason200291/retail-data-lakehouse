```python
# 03_product_kpi
# Product KPI Analysis

from pyspark.sql.functions import sum, round

clean_df = spark.table("clean_sales")

product_revenue = (
    clean_df
    .groupBy("Description")
    .agg(
        round(sum("Revenue"), 2).alias("TotalRevenue")
    )
    .orderBy("TotalRevenue", ascending=False)
)

display(product_revenue)
```

```python
from pyspark.sql.functions import sum

product_quantity = (
    clean_df
    .groupBy("Description")
    .agg(
        sum("Quantity").alias("TotalQuantity")
    )
    .orderBy("TotalQuantity", ascending=False)
)

display(product_quantity)
```

```python
from pyspark.sql.functions import count

product_transaction = (
    clean_df
    .groupBy("Description")
    .agg(
        count("Invoice").alias("TransactionCount")
    )
    .orderBy("TransactionCount", ascending=False)
)

display(product_transaction)
```

```python
product_revenue = (
    clean_df
    .groupBy("Description")
    .agg(
        round(sum("Revenue"), 2).alias("TotalRevenue")
    )
    .orderBy("TotalRevenue", ascending=False)
    .limit(10)
)

display(product_revenue)
```

```python
product_quantity = (
    clean_df
    .groupby("Description")
    .agg(
        sum("Quantity").alias("TotalQuantity")
    )
    .orderBy("TotalQuantity", ascending=False)
    .limit(10)
)


display(product_quantity)
```
